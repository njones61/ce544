# Pumping Test - Cooper-Jacob Analysis

A well fully penetrating a confined aquifer 20 m thick is pumped at a constant 10 L/s. The drawdown is recorded at an observation well 40 m away.

Cooper-Jacob: for $u = r^2S/(4Tt) < 0.05$, the drawdown is a straight line against log t:

$
s = \frac{q}{4\pi T}\ln\left(\frac{2.25\,Tt}{r^2S}\right), \qquad T = \frac{2.3\,q}{4\pi\,\Delta s}, \qquad S = \frac{2.25\,Tt_0}{r^2}
$

where $\Delta s$ is the change in drawdown over one log cycle of time and $t_0$ is the time where the line crosses s = 0.
